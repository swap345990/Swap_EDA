- complex number format  a+ib or a+jb

- where a= real number

- b= imaginary number

```python
num=10+20j
type(num)
```

```python
# apply dir on num
# see the methods
# real
# img
```

```python
num=10+20j
real=num.real
img=num.imag
conj=num.conjugate()

print(f"real value of {num} is {real}")
print(f"imaginary value of {num} is {img}")
print(f"conjugate value of {num} is {conj}")
```

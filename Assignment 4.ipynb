```python
import random
count_even=0
count_odd=0
i=1
while i<6:
    num=random.randint(20,70)
    if num%2==0:
        count_even=count_even+1
        print(f"the count of even is :{count_even}")
    else:
        count_odd=count_odd+1
        print(f"the count of odd is :{count_odd}")
    i=i+1
```

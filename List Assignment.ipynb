```python
#Qn 1
import random
l=[]
for i in range(5):
    User_input=random.randint(4,9)
    l.append(User_input)
def contains():
    for s in l:
        if s==5:
            
            return "list Contains five"
    return "not contains five"
five=contains()
cou=l.count(5)
sor=l[1:4]
sor.sort()
w=[j for j in l if j<5]
print(f"The list is {l}")    
print(f"A the lenth of list is : {len(l)}")
print(f"B the last item of list is : {l[len(l)-1]}")
print(f"C the reverse order of  list is : {l[::-1]}")
print(f"D {five}")
print(f"E the count of five is : {cou}")
print(f"F removed list is  : {sor}")
print(f"G less than five integer is : {w}")
```

```python
#Qn 2
a=[]
for e in range(20):
    User_input=random.randint(1,100)
    a.append(User_input)
Avg=sum(a)/20   
maxx=max(a)
minn=min(a)
a.sort()
second_high=a[len(a)-2:len(a)-1]
second_small=a[1]
even=[eve for eve in a if eve%2==0]
    
print(f"the list1 is : {a}")
print(f"the avarage number is : {Avg}")
print(f"the maximum number is : {maxx}")
print(f"the minimum number is : {minn}")
print(f"the second highest number is : {second_high}")
print(f"the second smallest number is : {second_small}")
print(f"the even number is : {even}")
```

```python
basic_sal=eval(input("enter the basic salary:   "))
def gross_sal():
    try:
        dearness_allow=basic_sal*40/100
        house_rent=basic_sal*20/100
        return basic_sal+dearness_allow+house_rent
    
    except Exception as e:
        print(e)
    
op=gross_sal()
print(f"gross salary is {op}")
```

```python
def marks():
    try:
        math=eval(input("enter the obtained mark of Math:  "))
        sc=eval(input("enter the obtained mark of Science:  "))
        eng=eval(input("enter the obtained mark of English:  "))
        mil=eval(input("enter the obtained mark of MIL:  "))
        his=eval(input("enter the obtained mark of History:  "))
        total_mark=math+sc+eng+mil+his
        per_mark=round(total_mark/500*100,2)
        
        print(f"aggregate of mark obtained : {total_mark}")
        print(f"percentage of mark obtained : {per_mark} percentage")
        
    except Exception as e:
        print(e)
        
marks()
        
```

```python
def five_digit():
    try:
        number=eval(input("enter the five digit number:  "))
        while number>0:
            last_num=number%10
            number=number//10

            print(last_num,end="")
            
    except Exception as e:
        print(e)
            
five_digit()
     
    
```
